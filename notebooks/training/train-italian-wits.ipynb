## Configuration

```python
from sm_sip.config import TrainingConfig

# Define training configurations
CONFIGS = {
    "10k-60t": TrainingConfig(
        lang="it",
        num_samples=10000,
        similarity_threshold=0.60,
        output_model_name="sigext-wits-it-10k-060t",
    ),
    "25k-60t": TrainingConfig(
        lang="it",
        num_samples=25000,
        similarity_threshold=0.60,
        output_model_name="sigext-wits-it-25k-060t",
    ),
    "25k-65t": TrainingConfig(
        lang="it",
        num_samples=25000,
        similarity_threshold=0.65,
        output_model_name="sigext-wits-it-25k-065t",
    ),
    "25k-70t": TrainingConfig(
        lang="it",
        num_samples=25000,
        similarity_threshold=0.70,
        output_model_name="sigext-wits-it-25k-070t",
    ),
}
```

## Training Loop

```python
from sm_sip.pipelines.training import train_sigext
from sm_sip.utils.gpu import clear_gpu_memory

for name, config in CONFIGS.items():
    print(f"\n{'='*60}")
    print(f"Training: {name}")
    print(f"{'='*60}")
    
    model, tokenizer = train_sigext(config)
    
    del model, tokenizer
    clear_gpu_memory()
    print(f"  {name} complete!")
```
